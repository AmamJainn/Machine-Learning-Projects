```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
cleaned_dataset[['age', 'bmi', 'children']] = scaler.fit_transform(
    cleaned_dataset[['age', 'bmi', 'children']]
)
```

```python
from scipy.stats import pearsonr
selected_features = [
    'age', 'bmi', 'children', 'is_female', 'is_smoker',
    'region_northwest', 'region_southeast', 'region_southwest',
    'bmi_category_Normal', 'bmi_category_Overweight', 'bmi_category_Obese'
]


correlations = {}

for feature in selected_features:
    corr = pearsonr(cleaned_dataset[feature], cleaned_dataset['charges'])[0]
    correlations[feature] = corr


correlation_df = pd.DataFrame(list(correlations.items()), columns=['Feature', 'Pearson Correlation'])


correlation_df.sort_values(by='Pearson Correlation', ascending=False)
```

```python
cat_features = [
    'is_female', 'is_smoker',
    'region_northwest', 'region_southeast', 'region_southwest',
    'bmi_category_Normal', 'bmi_category_Overweight', 'bmi_category_Obese'
]


from scipy.stats import chi2_contingency
import pandas as pd

# Define significance level
alpha = 0.05

# Bin the target variable
cleaned_dataset['charges_bin'] = pd.qcut(cleaned_dataset['charges'], q=4, labels=False)

# Run Chi-Square test
chi2_results = {}

for col in cat_features:
    contingency = pd.crosstab(cleaned_dataset[col], cleaned_dataset['charges_bin'])
    chi2_stat, p_val, _, _ = chi2_contingency(contingency)
    decision = 'Reject Null (Keep Feature)' if p_val < alpha else 'Accept Null (Drop Feature)'
    chi2_results[col] = {
        'chi2_statistic': chi2_stat,
        'p_value': p_val,
        'Decision': decision
    }

# Convert to DataFrame
chi2_df = pd.DataFrame(chi2_results).T
chi2_df = chi2_df.sort_values(by='p_value')
chi2_df
```

**model building**

```python
from sklearn.model_selection import train_test_split
```

```python
X = final_df.drop("charges",axis = 1)
y = final_df["charges"]
```

```python
x_train,x_test,y_train,y_test = train_test_split(X,y,train_size=0.8,random_state = 42)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(y_test,y_pred)
r2
```

```python
# adjusted r2 Score

number_of_test_samples = x_test.shape[0]
number_of_features = x_test.shape[1]

adjusted_r2 = 1 - (
    ((1 - r2) * (number_of_test_samples - 1)) /
    (number_of_test_samples - number_of_features - 1)
)

print(adjusted_r2)
```
